```python
import sympy as sym
import numpy as np
import torch
```

```python
x = sym.symbols('x')
display(sym.tanh(x).diff())
display(sym.atan(x).diff())
```

```python
x, y, z = sym.symbols('x y z')
f = y * sym.exp(x/y) - z
sym.hessian(f, (x, y, z))
```

```python
x, theta = sym.symbols(r'x \theta', real=True, pos=True)
f = sym.log(sym.exp(-x * y))
f.simplify()
```

```python
x = sym.Symbol('x', real=True)
sig = 1 / (1 + sym.exp(-x))
sig.diff().simplify()
```

```python
(sig * (1-sig)).simplify()
```

```python
m, n = sym.symbols('m n')
X = sym.MatrixSymbol('X', n, m)
w = sym.MatrixSymbol('w', m, 1)
y = sym.MatrixSymbol('y', n, 1)
lambd = sym.Symbol(r'\lambda', real=True)
b = sym.MatrixSymbol('b', n, 1)
sym.diff((X@w + b - y).T @ (X@w + b - y) + lambd * w.T @ w, b)
```

```python
x, y, z = sym.symbols('x y z', real=True)
f = y * sym.exp(x/y) - z
sym.hessian(f, (x, y, z))
```

```python
lambd = sym.Symbol(r'\lambda', real=True, positive=True)
x = sym.Symbol('x', real=True, positive=True)
y = sym.Symbol('y', real=True, positive=True)
f = sym.log(1 - sym.exp(-x)) * lambd
f.diff(x, 2).simplify()
```

```python
f = x**2/y
sym.hessian(f, (x, y))
```
